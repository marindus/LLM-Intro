```python
from google import genai
import os
from dotenv import load_dotenv
from google.genai import types
from openai import OpenAI
import anthropic
```

```python
load_dotenv()
# Access the API key using the variable name defined in the .env file
google_api_key = os.getenv("GOOGLE_API_KEY")
openai_api_key = os.getenv("OPENAI_API_KEY")
deepinfra_api_key = os.getenv("DEEPINFRA_API_KEY")
anthropic_api_key = os.getenv("ANTHROPIC_API_KEY")
```

## Google

https://ai.google.dev/gemini-api/docs/text-generation?hl=de
examples: https://colab.research.google.com/github/google-gemini/cookbook/blob/main/quickstarts/Get_started.ipynb?hl=de#scrollTo=mIC0pLnJdI8m

```python
# Erstellt einen Gemini-Client, so kann ich die API benutzen
client = genai.Client(api_key=google_api_key)

# Sendet eine Anfrage an ein bestimmtes Gemini-Modell
response = client.models.generate_content(
    model="gemini-2.5-flash", # Hier wird das Modell ausgewählt
    contents="What's the largest planet in our solar system?" # Der Prompt
)

# Gibt nur den Textteil der Modellantwort aus.
print(response.text)
```

```python

print(response.text)
```

```python
# Eine System Instruction sagt dem Modell, wie es sich grundsätzlich verhalten soll.
# In diesem Fall soll das Modell wie ein erfahrener Softwareentwickler und hilfreicher Coding-Assistent antworten.
system_instruction = """
  You are an expert software developer and a helpful coding assistant.
  You are able to generate high-quality code in any programming language.
"""
# Hier wird eine Konfiguration für die spätere Textgenerierung erstellt.
chat_config = types.GenerateContentConfig(
    system_instruction=system_instruction,
)

# Dieses Chat-Objekt kann später mehrere Nachrichten verarbeiten und dabei den Gesprächskontext behalten.
chat = client.chats.create(
    model="gemini-2.5-flash", # Hier wird das Modell ausgewählt
    config=chat_config, # Hier wird die oben erstellte Konfiguration an den Chat übergeben.
)
```

```python
# Sendet eine Nachricht an den zuvor erstellten Gemini-Chat.
response = chat.send_message("Write a function that checks if a year is a leap year.")
```

```python
# Gibt den generierten Antworttext des Modells aus.
response.text
```

# Openai
https://platform.openai.com/docs/guides/text

Use GPT-5.4-nano

```python
# Erstellt einen OpenAI-Client
client = OpenAI(api_key=openai_api_key)

# Erstellt eine neue Anfrage an die OpenAI Responses API.
response = client.responses.create(
    model="gpt-5.4-nano", # Hier wird das Modell ausgewählt
    reasoning={"effort": "low"}, # reasoning steuert, wie viel interne Denk-/Planungsarbeit das Modell machen soll. Es gibt (minimal, low, medium und high)
    input=[
        {
            # Die developer-Nachricht ist eine Anweisung an das Modell.
            "role": "developer",
            "content": "Talk like a pirate."
        },
        {
            # Die user-Nachricht ist die eigentliche Frage des Benutzers.
            "role": "user",
            "content": "Are semicolons optional in JavaScript?"
        }
    ]
)

# Gibt den generierten Antworttext des Modells aus.
print(response.output_text)
```

# Deepinfra
https://deepinfra.com/docs/openai_api

goal: meta-llama/Llama-3.3-70B-Instruct-Turbo


```python
# VON CHATGPT GENERIERT
# Erstellt einen Client für DeepInfra.
# DeepInfra ist OpenAI-kompatibel, deshalb können wir den OpenAI-Client wiederverwenden.
# Der Unterschied zu OpenAI ist die base_url.
deepinfra_client = OpenAI(
    api_key=deepinfra_api_key,
    base_url="https://api.deepinfra.com/v1/openai"
)
```

```python
# VON CHATGPT GENERIERT
# Sendet eine Chat-Anfrage an ein Llama-Modell über DeepInfra.
response = deepinfra_client.chat.completions.create(
    
    # Auswahl des Modells aus dem DeepInfra Model Catalog.
    model="meta-llama/Llama-3.3-70B-Instruct-Turbo",
    
    # messages enthält die Konversation.
    # system = Verhalten/Rolle des Modells
    # user = eigentliche Frage des Benutzers
    messages=[
        {
            "role": "system",
            "content": "You are a helpful coding assistant."
        },
        {
            "role": "user",
            "content": "Write a Python function that checks if a year is a leap year."
        }
    ]
)

# Gibt die reine Textantwort des Modells aus.
print(response.choices[0].message.content)
```

```python
# Erstellt einen API-Client, es wird DeepInfra verwendet da OpenAI-kompatible API anbietet um DeepInfra zu verwenden.
openai_client = OpenAI(
    api_key=deepinfra_api_key,
    base_url="https://api.deepinfra.com/v1/openai",
)
```

```python
# Sendet eine Chat-Anfrage an ein Modell, das über DeepInfra gehostet wird.
chat_completion = openai_client.chat.completions.create(
    model="meta-llama/Llama-4-Maverick-17B-128E-Instruct-FP8", # Hier wird das Modell ausgewählt. Dieses Modell läuft bei DeepInfra, nicht lokal auf deinem Computer.
    # messages enthält den bisherigen Gesprächsverlauf. Chatverlauf simulieren
    messages=[
        {"role": "system", "content": "Respond like a michelin starred chef."},
        {"role": "user", "content": "Can you name at least two different techniques to cook lamb?"},
        {"role": "assistant", "content": "Bonjour! Let me tell you, my friend, cooking lamb is an art form, and I'm more than happy to share with you not two, but three of my favorite techniques to coax out the rich, unctuous flavors and tender textures of this majestic protein. First, we have the classic \"Sous Vide\" method. Next, we have the ancient art of \"Sous le Sable\". And finally, we have the more modern technique of \"Hot Smoking.\""},
        {"role": "user", "content": "Tell me more about the second method."},
    ]
)
```

```python
# Gibt den Text der Modellantwort aus.
print(chat_completion.choices[0].message.content)
```

## Anthropic

https://docs.claude.com/en/docs/get-started#python

```python
# Erstellt einen Anthropic-Client. Dieser Client ist die Verbindung zwischen deinem Python-Code und der Anthropic API.
client = anthropic.Anthropic()

# Erstellt eine neue Nachricht bzw. Anfrage an Claude.
message = client.messages.create(
    model="claude-sonnet-4-5", # Hier wird das Claude-Modell ausgewählt.
    max_tokens=1000, # max_tokens begrenzt die maximale Länge der Antwort.
    messages=[
        {
            "role": "user", # "user" bedeutet: Das ist die Eingabe bzw. Frage des Benutzers.
            "content": "What should I search for to find the latest developments in renewable energy?" # Das ist der Prompt, der an Claude gesendet wird.
        }
    ]
)
```

```python
# Gibt den Text der ersten Antwort-Komponente aus.
print(message.content[0].text)
```
